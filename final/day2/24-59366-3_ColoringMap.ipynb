```python
graph = {
    'A': ['B', 'C', 'D'],
    'B': ['A', 'C', 'E'],
    'C': ['A', 'B', 'D', 'E'],
    'D': ['A', 'C'],
    'E': ['B', 'C']
}

colors = ['Red', 'Green', 'Blue']

assigned_colors = {}
```

```python
def is_valid(vertex, color):
    for neighbor in graph[vertex]:
        if assigned_colors.get(neighbor) == color:
            return False
    return True

def color_graph(vertices, index=0):
    
    if index == len(vertices):
        return True
    
    vertex = vertices[index]
    
    for color in colors:
        if is_valid(vertex, color):
            assigned_colors[vertex] = color
            
            if color_graph(vertices, index + 1):
                return True
            
            assigned_colors[vertex] = None
            

vertices_list = list(graph.keys())

if color_graph(vertices_list):
    print("AI Search Successful! Map colored as follows:")
    for v, c in assigned_colors.items():
        print(f"Vertex {v} -> {c}")
else:
    print("No solution found with the given colors.")
```
