```python
def f(x):
    return -x**2
```

```python
x = -5.0
step = 0.1

for i in range(1000):
    current_val = f(x)

    left_val = f(x - step)
    right_val = f(x + step)
    if left_val > current_val and left_val >= right_val:
        x = x - step

    elif right_val > current_val and right_val >= left_val:
        x = x + step
    else:
        break

print(f" Peak found at: x = {x:.4f}")
print(f" Peak found at: x = {f(x):.4f}")
```
