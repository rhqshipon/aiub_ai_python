## Setup and Data

```python
import random

city_names = ['A', 'B', 'C', 'D', 'E']

distances = [
    [0,  10, 15, 20, 12],  # A
    [10,  0, 35, 25, 18],  # B
    [15, 35,  0, 30, 22],  # C
    [20, 25, 30,  0, 16],  # D
    [12, 18, 22, 16,  0]   # E
]
```

## Evaluation Functions

```python
def calculate_distance(route):
    dist = distances[0][route[0]]
    for i in range(len(route) - 1):
        dist += distances[route[i]][route[i+1]]
    dist += distances[route[-1]][0]
    return dist

def calculate_fitness(route):
    return 1.0 / calculate_distance(route)
```

## Genetic Algorithm Core Operators

```python
def create_population(size):
    population = []
    for _ in range(size):
        route = [1, 2, 3, 4] # B, C, D, E
        random.shuffle(route) 
        population.append(route)
    return population

def selection(population, fitnesses):
    total_fitness = sum(fitnesses)
    pick = random.uniform(0, total_fitness)
    current = 0
    for i, f in enumerate(fitnesses):
        current += f
        if current > pick:
            return population[i]
    return population[-1]

def crossover(parent1, parent2):
    child = parent1[:2]
    for gene in parent2:
        if gene not in child:
            child.append(gene)
    return child

def mutate(route, mutation_rate=0.2):
    if random.random() < mutation_rate:
        idx1, idx2 = random.sample(range(len(route)), 2)
        route[idx1], route[idx2] = route[idx2], route[idx1]
    return route
```

## The Main Loop & Output

```python
# Parameters 
POPULATION_SIZE = 4
GENERATIONS = 3 

print("Genetic Algorithm\n")
population = create_population(POPULATION_SIZE)

for gen in range(GENERATIONS):
    print(f"************ GENERATION {gen + 1} ************")
    fitnesses = [calculate_fitness(ind) for ind in population]
    
    print("Current Population & Fitness:")
    for i, ind in enumerate(population):
        print(f"  Route: {ind} | Dist: {calculate_distance(ind)} | Fitness: {fitnesses[i]:.5f}")
        
    new_population = []
    
    for _ in range(POPULATION_SIZE // 2):
        p1 = selection(population, fitnesses)
        p2 = selection(population, fitnesses)
        print(f"\n  Selected Parents: {p1} & {p2}")
        
        c1, c2 = crossover(p1, p2), crossover(p2, p1)
        c1, c2 = mutate(c1), mutate(c2)
        
        print(f"  Offspring generated: {c1} & {c2}")
        new_population.extend([c1, c2])
        
    population = new_population
    print("\n")

print("************ FINAL RESULT ************")
fitnesses = [calculate_fitness(ind) for ind in population]
best_index = fitnesses.index(max(fitnesses))
best_route = population[best_index]

full_route_names = ['A'] + [city_names[i] for i in best_route] + ['A']

print(f"Best Route (City Names): {' -> '.join(full_route_names)}")
print(f"Minimum Total Distance:  {calculate_distance(best_route)}")
print(f"Best Fitness Value:      {max(fitnesses):.5f}")
```
